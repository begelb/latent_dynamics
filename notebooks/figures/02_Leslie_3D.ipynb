### Latent trajectories of the two 4-periodic orbits

Each orbit is encoded into the latent space and pushed forward four times under the latent map. The orbit inside the spurious Morse set never settles to a fixed point — that is the failure this example exposes.

```python
LESLIE3D_PERIODIC_PTS = {
    0: [
        [102.59382834, 4.62509476, 0.59276684],
        [6.47696572e-02, 7.18156798e01, 3.23756633e00],
        [1.20972812e00, 4.53387600e-02, 5.02709759e01],
        [6.60727793, 0.84680968, 0.03173713],
    ],
    1: [
        [20.09019989, 2.26201326, 21.10982997],
        [14.41254064, 14.06313992, 1.58340928],
        [43.08128567, 10.08877845, 9.84419795],
        [3.23144751, 30.15689997, 7.06214491],
    ],
}

exp_sp.show_latent_trajectory(LESLIE3D_PERIODIC_PTS, steps=4)
```
