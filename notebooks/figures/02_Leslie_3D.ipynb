# Section 5.2 — Three-dimensional Leslie: initial condition with two attractors

## What this notebook shows

The three-generation **Leslie population model** (paper §5.2), a genuinely three-dimensional system, studied through a **two-dimensional** latent model. This notebook places a cautionary tale and a success story side by side:

- **Example 1 — a *spurious* attractor (§5.2.1).** The latent Morse graph comes out *tristable* (three minimal nodes), but the true system has only two stable period-four orbits — one latent attractor corresponds to no attractor of the real dynamics. This is not just numerical noise: the semiconjugacy-error bound required by the main theorem is *violated* at that node. Lesson: a small training loss does **not** by itself guarantee that latent attractors are real.
- **Example 2 — bistability correctly identified (§5.2.2).** At other parameters the latent Morse graph has two minimal nodes whose Conley indices $(x^4-1,0,0)$ are those of period-four orbits, faithfully matching the true system.

## Example 1 — spurious attractor (§5.2.1)

```python
from latentdynamics.replay import load_experiment

exp_sp = load_experiment("leslie3d_example1_replay")
exp_sp
```

```python
exp_sp.show_morse_graph()
```

```python
exp_sp.show_morse_sets()
```

### Latent trajectories of the two 4-periodic orbits

Each orbit is encoded into the latent space and pushed forward four times under the latent map. The orbit inside the spurious Morse set never settles to a fixed point — that is the failure this example exposes.
