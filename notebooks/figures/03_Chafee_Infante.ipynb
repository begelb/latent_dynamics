# Section 5.3 — Chafee-Infante PDE with bistability

## What this notebook shows

The **Chafee–Infante** semilinear parabolic PDE $u_t = u_{xx} + \lambda(u - u^3)$ on $[0,\pi]$ with Dirichlet boundary conditions (paper §5.3). At $\lambda = 28$ the equation has eleven steady states — **two stable**, nine unstable. Discretizing in space yields a **64-dimensional** system; we learn a two-dimensional latent model and compute its Morse graph, which recovers the **bistability** (the two stable steady states $u_\pm$) as two minimal nodes.
