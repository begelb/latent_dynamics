# Section 5.1 — Two-dimensional Leslie with contracting dynamics

## What this notebook shows

A *warm-up* for the method (paper §5.1). We take a two-dimensional nonlinear **Leslie population map** — which has a period-doubling cascade, an attracting invariant circle, and the Newhouse phenomenon, so its dynamics resist rigorous characterization even when the map is known — and embed it in a **ten-dimensional** phase space with strongly contracting dynamics in the eight extra coordinates ($\bar\theta = (23.5, 23.5)$).

From a deliberately **sparse** dataset (20 trajectories), we learn an autoencoder (encoder $E$, decoder $D$) together with a latent dynamics map $g$ that form an *$\epsilon$-approximate semiconjugacy* to the full system. We then run rigorous combinatorial-topological machinery (CMGDB) on $g$ in the low-dimensional latent space and read off the global dynamics as a **Morse graph**.

The Morse graph below has **two minimal nodes — i.e. bistability** — recovering both attractors of the planar map: a stable invariant circle (Conley index $(x-1,\,x-1,\,0)$) and a stable period-six orbit (Conley index $(x^6-1,\,0,\,0)$).
